```python
import pandas as pd
```

```python
scene_URL = "https://media.githubusercontent.com/media/zoepierrat/ECOSTRESS_C2_L3_ET_Validation/refs/heads/main/data/eco_cal_val_data/2025-03-03_ECOSTRESS_calval_C2_merged.csv"
scene_filename = "../ECOv002_calval_tables/orbit_scene_tile.csv"
```

```python
scene_df = pd.read_csv(scene_URL)
scene_df = scene_df[["tower", "orbit", "scene", "tile", "time_UTC"]]
scene_df = scene_df.drop_duplicates()
scene_df = scene_df.dropna()
scene_df["orbit"] = scene_df["orbit"].astype(int)
scene_df["scene"] = scene_df["scene"].astype(int)
scene_df.rename(columns={"tower": "ID"}, inplace=True)
scene_df
```

```python
scene_df.to_csv(scene_filename, index=False)
```
